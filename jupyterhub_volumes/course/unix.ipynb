```bash
# your work here
# 1. list /etc
# 2. redirect output to listing.txt
# 3. show first five lines
```

## Exercise 3 – Using a For Loop

Write a loop that prints the line count of each `.txt` file in the current directory.

```bash
# your work here
# hint: use wc -l
```

```bash
# Solution
for f in *.txt; do
  echo "$f: $(wc -l < "$f") lines"
done
```
